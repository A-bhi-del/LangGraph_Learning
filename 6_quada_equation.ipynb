```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class QuadaState(TypedDict):
    a: int
    b: int
    c: int

    equ: str
    dis: float
    res: str
```

```python
def show_equation(state: QuadaState) -> QuadaState:
    equation = f'{state["a"]}x2{state["b"]}x{state["c"]} = 0'

    return {'equ': equation}

def calcu_discriminent(state: QuadaState) -> QuadaState:
    dis = state['b']**2 - 4*state['a']*state['c']

    return {'dis': dis}

def real_roots(state: QuadaState) -> QuadaState:
    root1 = (-state['b'] + state['dis']**0.5)/(2*state['a'])
    root2 = (-state['b'] - state['dis']**0.5)/(2*state['a'])

    result = f'The roots are {root1} and {root2}'

    return {'res': result}

def repeated_roots(state: QuadaState)-> QuadaState:
    root = -state['b']/(2*state['a'])

    result = f'Only repeated roots are {root}'

    return {'res': result}

def no_real_roots(state: QuadaState) -> QuadaState:
    result = f'There are no real roots'

    return {'res': result}

def check_condition(state: QuadaState) -> Literal["real_roots", "repeated_roots", "no_real_roots"]:
    if state['dis'] > 0:
        return "real_roots"
    elif state['dis'] == 0:
        return "repeated_roots"
    else:
        return "no_real_roots"
```

```python
graph = StateGraph(QuadaState)

graph.add_node('show_equation', show_equation)
graph.add_node('calcu_discriminent', calcu_discriminent)
graph.add_node("real_roots", real_roots)
graph.add_node("repeated_roots", repeated_roots)
graph.add_node("no_real_roots", no_real_roots)

graph.add_edge(START, "show_equation")
graph.add_edge("show_equation", "calcu_discriminent")

graph.add_conditional_edges("calcu_discriminent", check_condition)
graph.add_edge("real_roots", END)
graph.add_edge("repeated_roots", END)
graph.add_edge("no_real_roots", END)

workflow = graph.compile()
```

```python
initial_state = {'a': 1, 'b': 2, 'c': 3}

final_state = workflow.invoke(initial_state)

print(final_state)
```
