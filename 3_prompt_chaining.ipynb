```python
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
model = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
    score: float
```

```python
def create_outline(state: BlogState) -> BlogState:
    title = state['title']

    prompt = f'Generate a detail outline for the topic - {title}'

    outline = model.invoke(prompt).content[0]['text']

    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f'write a detail blog on topic - {title} with the given outline \n {outline}'

    content = model.invoke(prompt).content[0]['text']

    state['content'] = content

    return state
```

```python
def eveluate_blog(state: BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']
    content = state['content']

    prompt = f'according to the following outline \n {outline} \n and content \n {content} \n what is the score of the blog on topic - {title}'

    score = model.invoke(prompt).content[0]['text']

    state['score'] = score

    return state
```

```python
graph = StateGraph(BlogState)

graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)
graph.add_node('eveluate_blog', eveluate_blog)

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', 'eveluate_blog')
graph.add_edge('eveluate_blog', END)

workflow = graph.compile()
```

```python
initial_state = {'title': 'Python Programming'}
final_state = workflow.invoke(initial_state)

# print(final_state)
print(final_state['score'])
```
